````python
import feedparser
from urllib.parse import quote
import pandas as pd
````

````python
def get_google_news(query, language="id", country="ID"):
    encoded_query = quote(query)

    url = (
        f"https://news.google.com/rss/search?"
        f"q={encoded_query}&hl={language}&gl={country}&ceid={country}:{language}"
    )

    feed = feedparser.parse(url)

    articles = []

    for entry in feed.entries:
        articles.append({
            "title": entry.get("title", ""),
            "url": entry.get("link", ""),
            "published": entry.get("published", ""),
            "summary": entry.get("summary", "")
        })

    return articles
````

````python
articles = get_google_news("properti Indonesia")

print(f"Retrieved {len(articles)} articles")

for article in articles[:5]:
    print("\nTITLE:", article["title"])
    print("DATE:", article["published"])
    print("URL:", article["url"])
````

````python
df = pd.DataFrame(articles)

df.head()
````

````python
queries = [
    "properti Indonesia",
    "KPR Indonesia",
    "harga rumah Indonesia",
    "suku bunga KPR",
    "pasar properti Indonesia"
]

all_articles = []

for query in queries:
    results = get_google_news(query)

    for article in results:
        article["query"] = query
        all_articles.append(article)

news_df = pd.DataFrame(all_articles)

print(f"Total articles retrieved: {len(news_df)}")
````

````python
news_df.head()
````

````python
news_df = news_df.drop_duplicates(subset="url")

print(f"After removing duplicates: {len(news_df)}")
````

````python
news_df[["title", "published", "query"]].head(20)
````

````python
from bs4 import BeautifulSoup

def clean_html(text):
    return BeautifulSoup(text, "html.parser").get_text(" ", strip=True)

news_df["summary"] = news_df["summary"].apply(clean_html)

news_df.head()
````

````python
news_df["published_dt"] = pd.to_datetime(
    news_df["published"],
    errors="coerce",
    utc=True
)

news_df = news_df.sort_values(
    "published_dt",
    ascending=False
)

candidate_df = news_df.head(20).copy()

print(f"Sending {len(candidate_df)} articles to Gemini")
````

````python
candidate_df[["title", "published", "query"]]
````

````python
selection_prompt = """
You are the editorial selection engine for an Indonesian
property marketplace.

The target audience is Indonesian young adults aged 22–28,
especially people who are beginning to earn their own income
and may eventually rent or buy a home.

Below is a list of recent news articles.

Your task is to select the TOP 5 articles that have the
highest potential for an educational Instagram post.

Evaluate each article based on:

1. Relevance to Indonesian property and housing
2. Geographic relevance to Indonesia
3. Practical usefulness to young adults
4. Financial or policy implications
5. Presence of concrete facts, numbers, or developments
6. Timeliness
7. Potential to explain the topic clearly in an Instagram carousel

Avoid or heavily penalize:

- Hoaxes or misinformation
- Job scams
- Pure corporate announcements
- Promotional advertisements
- Company marketing content
- Celebrity/gossip stories
- News unrelated to housing or property
- Stories with little educational value

Important:
Do not invent information.
Base your evaluation only on the information provided.

Geographic rule:
Prioritize news directly concerning Indonesia.

International news should normally be excluded unless it provides
a highly useful comparison, benchmark, or lesson that can be
directly connected to Indonesian housing consumers.

Return ONLY valid JSON in this format:

{
  "ranked_articles": [
    {
      "rank": 1,
      "article_index": 0,
      "title": "...",
      "score": 0,
      "topic": "...",
      "reason": "..."
    }
  ]
}

Return exactly 5 articles, ranked from best to worst.
"""
````

````python
articles_for_ai = []

for index, row in candidate_df.iterrows():
    articles_for_ai.append({
        "index": int(index),
        "title": row["title"],
        "published": row["published"],
        "summary": row["summary"],
        "url": row["url"]
    })

articles_text = "\n\n".join(
    [
        f"""
ARTICLE INDEX: {article['index']}
TITLE: {article['title']}
PUBLISHED: {article['published']}
SUMMARY: {article['summary']}
URL: {article['url']}
"""
        for article in articles_for_ai
    ]
)
````

````python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents=selection_prompt + "\n\n" + articles_text
)

print(response.text)
````

````python
import json
selection_result = json.loads(response.text.strip('```json').strip('```'))
selection_result
````

````python
winner_index = selection_result["ranked_articles"][0]["article_index"]

winner = candidate_df.loc[winner_index]

print("Selected article:")
print(winner["title"])
print(winner["url"])
````

````python
generation_prompt = """
You are a social media content strategist for an Indonesian
online property marketplace.

Your target audience is Indonesian Gen Z and young millennials,
approximately 22–28 years old, who are beginning to manage
their own finances and may be considering renting or buying
their first home.

Using ONLY the supplied news article, create an educational
Instagram carousel about the story.

The goal is not simply to summarize the article.
The goal is to explain why the development matters to a young
Indonesian who may eventually buy a home or take a KPR.

Create exactly 7 slides:

Slide 1:
A strong hook that makes the audience want to keep reading.

Slide 2:
Explain what happened in simple language.

Slide 3:
Explain the most important number, statistic, or development.

Slide 4:
Explain why this matters for the property market.

Slide 5:
Explain what this could mean for a young prospective
homebuyer.

Slide 6:
Give a practical financial takeaway.
Do not give personalized financial advice.

Slide 7:
Give a concise takeaway and engaging CTA.

Tone:
- conversational
- intelligent
- accessible
- slightly playful
- appropriate for Indonesian Gen Z
- not overly corporate
- not excessively slang-heavy

IMPORTANT FACTUAL RULES:
- Do not invent statistics.
- Do not invent quotes.
- Do not invent dates.
- Do not invent financial advice or claims.
- Every factual claim must be supported by the supplied article.
- Clearly distinguish between facts from the article and general
  financial guidance.
- If the article does not provide enough information to support
  a claim, do not make that claim.

Also create:
1. An Instagram caption of approximately 100–150 words.
2. A short CTA.
3. Source attribution.

Return ONLY valid JSON in this format:

{
  "carousel": [
    {
      "slide": 1,
      "headline": "...",
      "body": "..."
    }
  ],
  "caption": "...",
  "cta": "...",
  "source": "..."
}
"""
````
